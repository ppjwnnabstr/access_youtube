```python
import os
import yt_dlp
from google.colab import files

# youtube_url = "https://youtu.be/wjaN-dhrxzY?si=W8AdcMbM1q-61Jul"
# youtube_url = "https://youtu.be/SkxukBD45g0?si=Olf13z5jL_c5D3Ky"
youtube_url = "https://youtu.be/UMxAyoJfDL8?si=gLezrjDoTj4sJ5LX"

output_dir = "/content/mp4"
os.makedirs(output_dir, exist_ok=True)

ydl_opts = {
    "format": "bestvideo+bestaudio/best",
    "outtmpl": f"{output_dir}/%(title)s.%(ext)s",
    "noplaylist": True,
    "merge_output_format": "mp4",
}

try:
    print("กำลังดาวน์โหลดวิดีโอ...")

    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        info = ydl.extract_info(
            youtube_url,
            download=True
        )

        title = info.get("title", "output")

    # หาไฟล์ MP4 ที่สร้างขึ้น
    mp4_files = [
        os.path.join(output_dir, f)
        for f in os.listdir(output_dir)
        if f.lower().endswith(".mp4")
    ]

    if not mp4_files:
        raise Exception("ไม่พบไฟล์ MP4 หลังจากดาวน์โหลด")

    # เลือกไฟล์ล่าสุด
    mp4_file = max(
        mp4_files,
        key=os.path.getctime
    )

    print("\nดาวน์โหลดวิดีโอจาก YouTube เสร็จแล้ว!")
    print(f"ชื่อวิดีโอ: {title}")
    print(f"ไฟล์ใน Colab: {mp4_file}")

    # ดาวน์โหลดไฟล์ลงเครื่องอัตโนมัติ
    print("\nกำลังเริ่มดาวน์โหลดไฟล์ลงเครื่อง...")
    files.download(mp4_file)

except Exception as e:
    print("\nเกิดข้อผิดพลาด:")
    print(e)
```

```python
import os
import yt_dlp
from google.colab import files

youtube_url = "https://youtu.be/UMxAyoJfDL8?si=gLezrjDoTj4sJ5LX"

output_dir = "/content/mp3"
os.makedirs(output_dir, exist_ok=True)

ydl_opts = {
    # ดาวน์โหลด audio ที่ดีที่สุด
    "format": "bestaudio/best",

    # ตั้งชื่อไฟล์
    "outtmpl": f"{output_dir}/%(title)s.%(ext)s",

    # ไม่เอา playlist
    "noplaylist": True,

    # แปลงเป็น MP3
    "postprocessors": [
        {
            "key": "FFmpegExtractAudio",
            "preferredcodec": "mp3",
            "preferredquality": "192",
        }
    ],
}

try:
    print("กำลังดาวน์โหลด Audio...")

    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        info = ydl.extract_info(
            youtube_url,
            download=True
        )

        title = info.get("title", "output")

    # หาไฟล์ MP3 ที่สร้างขึ้น
    mp3_files = [
        os.path.join(output_dir, f)
        for f in os.listdir(output_dir)
        if f.lower().endswith(".mp3")
    ]

    if not mp3_files:
        raise Exception("ไม่พบไฟล์ MP3 หลังจากดาวน์โหลด")

    # เลือกไฟล์ล่าสุด
    mp3_file = max(
        mp3_files,
        key=os.path.getctime
    )

    print("\nดาวน์โหลด Audio จาก YouTube เสร็จแล้ว!")
    print(f"ชื่อวิดีโอ: {title}")
    print(f"ไฟล์ใน Colab: {mp3_file}")

    # ดาวน์โหลดไฟล์ลงเครื่อง
    print("\nกำลังดาวน์โหลด MP3 ลงเครื่อง...")
    files.download(mp3_file)

except Exception as e:
    print("\nเกิดข้อผิดพลาด:")
    print(e)
```
